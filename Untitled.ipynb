```python
import pandas as pd
import numpy as np
```

```python
file_path = r"C:\Users\npdan\IH_LABS/lab-tableau-advanced/marketing_customer_value_analysis.csv"
df = pd.read_csv(file_path)
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.regplot(
    data=df,
    x='Income',
    y='Customer Lifetime Value',
    scatter_kws={'alpha':0.5},
    line_kws={'color':'red'}
)

plt.title('Customer Lifetime Value vs. Income')
plt.xlabel('Income')
plt.ylabel('Customer Lifetime Value')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
pearson_corr = df['Income'].corr(df['Customer Lifetime Value'], method='pearson')
spearman_corr = df['Income'].corr(df['Customer Lifetime Value'], method='spearman')

print(f"Pearson correlation: {pearson_corr:.2f}")
print(f"Spearman correlation: {spearman_corr:.2f}")
```

```python
#Interpretation :
    # - no linear relationship between Income and CLV
    # - Widely scattered points → weak correlation
    # - Wide band: More uncertainty
    # - Pearson and Spearman scores are both very low : it's a confirmation that there is no significant correlation between Income and CLT
```
